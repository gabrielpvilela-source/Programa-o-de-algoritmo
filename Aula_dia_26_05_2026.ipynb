```python
# Calculadora com as 4 operações

# menu
# Opção 1 - Somar
# Opção 2 - Subtrair
# Opção 3 - Multiplicar
# Opção 4 - Divisão
# Opção 5 - Sair

print ("--------MENU---------")
print ("")

print ("Opção 1 - Somar")
print ("Opção 2 - Subtrair")
print ("Opção 3 - Multiplicar")
print ("Opção 4 - Divisão")
print ("Opção 5 - Sair")
print ("----------------------")

opcao_para_loop = 0

while opcao_para_loop != 5:
  opcao_lida = 0
  num_a_lido = 0.0
  num_b_lido = 0.0

  try:
    opcao_str_input = input("Digite a opção desejada: ")
    opcao_lida = int(opcao_str_input)
  except ValueError:
    print("Opção inválida. Por favor, digite um número de 1 a 5.")


  if opcao_lida == 5:
    print("Saindo da calculadora. Até logo!")
    opcao_para_loop = 5
  elif 1 <= opcao_lida <= 4:

    try:
      num_a_str_input = input("Digite o primeiro número: ")
      num_b_str_input = input("Digite o segundo número: ")
      num_a_lido = float(num_a_str_input)
      num_b_lido = float(num_b_str_input)


      print("")

      if opcao_lida == 1:
        print (f"O total da soma de {num_a_lido} + {num_b_lido} é igual a {num_a_lido + num_b_lido}")
      elif opcao_lida == 2:
        print (f"O total da subtração de {num_a_lido} - {num_b_lido} é igual a {num_a_lido - num_b_lido}")
      elif opcao_lida == 3:
        print (f"O total da multiplicação de {num_a_lido} * {num_b_lido} é igual a {num_a_lido * num_b_lido}")
      elif opcao_lida == 4:
        if num_b_lido == 0:
          print("Erro: Divisão por zero não é permitida.")
        else:
          print (f"O total da divisão de {num_a_lido} / {num_b_lido} é igual a {num_a_lido / num_b_lido}")
      print("")
    except ValueError:
      print("Entrada inválida. Por favor, digite números válidos.")
  else:
    if opcao_lida != 0:

      print ("Digite uma opção de 1 até 4 ou 5 para sair.")

  if opcao_para_loop != 5:
    print("")
```

```python
# Jogo inteligente

# 1 - par/ímpar
# 2 - Positivo
# 3 - Maior de 2 números
# 4 - Sair
```

```python
# Jogo inteligente

print ("\n--------JOGO INTELIGENTE---------")
print ("\nOpção 1 - Par/Ímpar")
print ("Opção 2 - Positivo")
print ("Opção 3 - Maior de 2 números")
print ("Opção 4 - Sair")
print ("---------------------------------")

jogo_ativo = 0

while jogo_ativo != 4:
  opcao_jogo_lida = 0

  try:
    opcao_jogo_str_input = input("Digite a opção desejada para o Jogo Inteligente: ")
    opcao_jogo_lida = int(opcao_jogo_str_input)
  except ValueError:
    print("Opção inválida. Por favor, digite um número de 1 a 4.")

  if opcao_jogo_lida == 4:
    print("Saindo do Jogo Inteligente. Até logo!")
    jogo_ativo = 4
  elif 1 <= opcao_jogo_lida <= 3:
    if opcao_jogo_lida == 1:
      try:
        numero = int(input("Digite um número inteiro para verificar se é par ou ímpar: "))
        if numero % 2 == 0:
          print(f"O número {numero} é PAR.")
        else:
          print(f"O número {numero} é ÍMPAR.")
      except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro válido.")
    elif opcao_jogo_lida == 2:
      try:
        numero = float(input("Digite um número para verificar se é positivo: "))
        if numero > 0:
          print(f"O número {numero} é POSITIVO.")
        elif numero < 0:
          print(f"O número {numero} é NEGATIVO.")
        else:
          print(f"O número {numero} é ZERO.")
      except ValueError:
        print("Entrada inválida. Por favor, digite um número válido.")
    elif opcao_jogo_lida == 3:
      try:
        num1 = float(input("Digite o primeiro número: "))
        num2 = float(input("Digite o segundo número: "))
        if num1 > num2:
          print(f"O maior número é {num1}.")
        elif num2 > num1:
          print(f"O maior número é {num2}.")
        else:
          print(f"Os números são iguais: {num1}.")
      except ValueError:
        print("Entrada inválida. Por favor, digite números válidos.")
  else:
    if opcao_jogo_lida != 0:
      print ("Digite uma opção de 1 até 3 ou 4 para sair.")

  if jogo_ativo != 4:
    print("\n---------------------------------")
```
