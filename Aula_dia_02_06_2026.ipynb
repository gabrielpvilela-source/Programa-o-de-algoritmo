```python
# Generate a random secret number between 0 and 10
secret_number = random.randint(0, 10)

attempts = 0
max_attempts = 3

print("Bem-vindo ao jogo de adivinhação!")
print(f"Estou pensando em um número entre 0 e 10. Você tem {max_attempts} tentativas.")

while attempts < max_attempts:
    try:
        # Get user's guess
        user_guess = int(input("Digite seu palpite (0-10): "))

        # Validate guess range
        if not (0 <= user_guess <= 10):
            print("Por favor, digite um número entre 0 e 10.")
            continue

        attempts += 1

        if user_guess == secret_number:
            print(f"Parabéns! Você adivinhou o número secreto {secret_number} em {attempts} tentativas!")
            break
        else:
            difference = abs(secret_number - user_guess)
            if difference == 1:
                print("Seu palpite está quente!")
            elif difference == 2:
                print("Seu palpite está morno.")
            elif difference >= 3:
                print("Seu palpite está frio.")

            if attempts < max_attempts:
                print(f"Você tem {max_attempts - attempts} tentativas restantes.")
            else:
                print("Suas tentativas acabaram.")

    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

if user_guess != secret_number and attempts == max_attempts:
    print(f"Game over! O número secreto era {secret_number}.")
```
